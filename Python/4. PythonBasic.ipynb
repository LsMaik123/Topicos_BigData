3. Loops

```python
palabras = cadena_1.split()
palabras
```

```python
for palabra in palabras:
    print(palabra)
```

```python
#Colocar Numeración a cada palabra
for idx, palabra in enumerate(palabras):
    print(idx, palabra)
```
